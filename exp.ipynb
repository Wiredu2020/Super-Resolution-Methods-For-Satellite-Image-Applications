```python
import pandas as pd
from types import SimpleNamespace

history_csv_path = 'training_history_lsman_x2.csv'  # change to your CSV file path

df = pd.read_csv(history_csv_path, index_col=0)
history_dict = {col: df[col].dropna().tolist() for col in df.columns}

lsman = SimpleNamespace(history=history_dict)

print("Loaded history keys:", list(history_dict.keys()))
```

```python
df.head()  # Display the first few rows of the DataFrame
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.plot(df.index, df['loss'], label='loss', marker='o')
plt.plot(df.index, df['val_loss'], label='val_loss', marker='o')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.title('Model Loss')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(df.index, df['keras_psnr'], label='keras_psnr', marker='o')
plt.plot(df.index, df['val_keras_psnr'], label='val_keras_psnr', marker='o')
plt.xlabel('Epoch')
plt.ylabel('PSNR')
plt.legend()
plt.title('Model PSNR')
plt.grid(True)

plt.tight_layout()
plt.show()
```
